```python
import pandas as pd
from geopy.distance import geodesic
import folium
```

```python
# Dataset
data = pd.read_csv("indianspots.csv")
dataset = pd.DataFrame(data)
```

```python
CATEGORIES = {

    "parks": "parks",
    "museum": "museum",
    "waterfalls": "waterfalls",
    "beaches": "beaches",
    "zoo": "zoo",
    "temples": "temple",
    "hills": "hill",
    "attraction": "tourist attraction",
}
```

```python
user_location = (12.133902, 78.1568531)  
radius = 10  # Radius in kilometers
category_filter = "tourist attraction"  # Example category

```

```python
def filter_by_distance(df, user_location, radius):
    filtered_data = []
    for _, row in df.iterrows():
        spot_location = (row['Latitude'], row['Longitude'])
        distance = geodesic(user_location, spot_location).km
        print(f"Checking {row['Name']}: Distance = {distance:.2f} km")  # Debug
        if distance <= radius:
            filtered_data.append({
                "Name": row['Name'],
                "Latitude": row['Latitude'],
                "Longitude": row['Longitude'],
                "Rating": row['Rating'],
                "Categories": row['Categories'],
                "Distance": distance
            })
    print(f"Number of spots within radius: {len(filtered_data)}")  # Debug
    return pd.DataFrame(filtered_data)


```

```python

def filter_by_category(df, category_filter):
    print(f"Applying category filter: {category_filter}")  # Debug
    if category_filter.lower() != "allcategories":
        category = CATEGORIES.get(category_filter.lower(), None)
        if category:
            df = df[df['Categories'].str.contains(category, case=False, na=False)]
    print(f"Filtered by category: {len(df)} spots remaining")  # Debug
    return df
```

```python
# Sort results
def sort_results(df):
    return df.sort_values(by=['Rating', 'Distance'], ascending=[False, True])

```

```python

def recommend_spots(user_location, radius, category_filter, dataset):
    print("Starting recommendation process...")  # Debug
    filtered_df = filter_by_distance(dataset, user_location, radius)
    print(f"Filtered by distance: {len(filtered_df)} spots remaining")  # Debug
    filtered_df = filter_by_category(filtered_df, category_filter)
    print(f"Filtered by category: {len(filtered_df)} spots remaining")  # Debug
    sorted_df = sort_results(filtered_df)
    print("Final sorted recommendations:")
    print(sorted_df)  # Debug
    return sorted_df
```

```python
import requests
from folium import PolyLine

def get_route(user_location, destination):
    """Fetch the shortest route using OSRM."""
    osrm_url = (
        f"http://router.project-osrm.org/route/v1/driving/"
        f"{user_location[1]},{user_location[0]};{destination[1]},{destination[0]}?overview=full&geometries=geojson"
    )
    response = requests.get(osrm_url)
    if response.status_code == 200:
        data = response.json()
        coordinates = data['routes'][0]['geometry']['coordinates']  # Extract the route
        # Convert coordinates to folium-friendly format (lat, lon pairs)
        return [(coord[1], coord[0]) for coord in coordinates]
    else:
        print(f"Error fetching route: {response.status_code}")
        return []

def visualize_map_with_route(user_location, recommendations):
    """Visualize the map with user location, recommendations, and route."""
    tourist_map = folium.Map(location=user_location, zoom_start=12)
    
    # Add user's location
    folium.Marker(
        location=user_location,
        popup="Your Location",
        icon=folium.Icon(color='blue')
    ).add_to(tourist_map)
    
    # Add recommended spots and routes
    for _, row in recommendations.iterrows():
        destination = (row['Latitude'], row['Longitude'])
        route = get_route(user_location, destination)
        
        # Add a marker for the destination
        folium.Marker(
            location=destination,
            popup=f"<b>{row['Name']}</b><br>Rating: {row['Rating']}<br>"
                  f"Distance: {row['Distance']:.2f} km",
            icon=folium.Icon(color='green')
        ).add_to(tourist_map)
        
        # Add the route to the map
        if route:
            folium.PolyLine(route, color="blue", weight=3, opacity=0.7).add_to(tourist_map)
    
    return tourist_map
```

```python
# Get recommendations
recommendations = recommend_spots(user_location, radius, category_filter, dataset)
if recommendations.empty:
    print("No spots match your criteria. Try increasing the radius or changing the category.")
else:
    print(recommendations)
```

```python
# Generate map visualization
if not recommendations.empty:
    tourist_map = visualize_map_with_route(user_location, recommendations)
    tourist_map.save("tourist_recommendations.html")
    print("Map saved as 'tourist_recommendations.html'")
```
